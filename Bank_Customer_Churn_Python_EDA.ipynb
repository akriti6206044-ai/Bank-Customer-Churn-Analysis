# 🏦 Bank Customer Churn Analysis — Python EDA\n\nThis notebook performs **Exploratory Data Analysis (EDA)** on the cleaned Bank Customer Churn dataset.\n\n**Tools:** Python, Pandas, NumPy, Matplotlib\n\n**Goal:** Understand customer churn patterns across geography, age, gender, credit score, tenure, products, balance, credit-card ownership, and active membership.\n

```python
# If needed, install packages first:\n# pip install pandas numpy matplotlib openpyxl\n\nimport pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\n\npd.set_option('display.max_columns', None)\npd.set_option('display.float_format', lambda x: f'{x:,.2f}')\nprint('Libraries imported successfully.')
```

## 1. Load the Dataset\n\nThe Excel workbook contains a `Customer_Data` sheet. Update the path if your Excel file is stored somewhere else.

```python
file_path = 'Bank_Customer_Churn_Cleaned.xlsx'\ndf = pd.read_excel(file_path, sheet_name='Customer_Data')\n\nprint(f'Dataset loaded: {df.shape[0]:,} rows × {df.shape[1]} columns')\ndf.head()
```

## 2. Understand the Dataset

```python
print('Shape:', df.shape)\nprint('\nColumns:')\nprint(df.columns.tolist())\n\nprint('\nData types:')\nprint(df.dtypes)
```

```python
df.info()
```

## 3. Data Quality Checks

```python
# Missing values\nmissing = df.isnull().sum().sort_values(ascending=False)\nprint('Missing values by column:')\nprint(missing)\n\n# Duplicate rows\nprint(f'\nDuplicate rows: {df.duplicated().sum()}')\n\n# Duplicate customer IDs\nprint(f'Duplicate CustomerId values: {df["CustomerId"].duplicated().sum()}')
```

```python
# Descriptive statistics for numeric columns\ndf.describe().T
```

## 4. Overall Churn KPI

```python
total_customers = len(df)\nchurned_customers = int(df['Exited'].sum())\nretained_customers = total_customers - churned_customers\nchurn_rate = df['Exited'].mean() * 100\n\nprint(f'Total Customers   : {total_customers:,}')\nprint(f'Churned Customers : {churned_customers:,}')\nprint(f'Retained Customers: {retained_customers:,}')\nprint(f'Churn Rate        : {churn_rate:.2f}%')
```

## 5. Churn Distribution

```python
status_counts = df['Exited'].map({0: 'Retained', 1: 'Churned'}).value_counts()\nprint(status_counts)\n\nstatus_counts.plot(kind='bar')\nplt.title('Customer Churn Distribution')\nplt.xlabel('Customer Status')\nplt.ylabel('Number of Customers')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 6. Churn by Geography

```python
geo_summary = df.groupby('Geography').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\ngeo_summary['churn_rate'] *= 100\ngeo_summary.sort_values('churn_rate', ascending=False)
```

```python
geo_summary.plot(x='Geography', y='churn_rate', kind='bar', legend=False)\nplt.title('Churn Rate by Geography')\nplt.xlabel('Geography')\nplt.ylabel('Churn Rate (%)')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 7. Churn by Gender

```python
gender_summary = df.groupby('Gender').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\ngender_summary['churn_rate'] *= 100\ngender_summary.sort_values('churn_rate', ascending=False)
```

```python
gender_summary.plot(x='Gender', y='churn_rate', kind='bar', legend=False)\nplt.title('Churn Rate by Gender')\nplt.xlabel('Gender')\nplt.ylabel('Churn Rate (%)')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 8. Age Analysis

```python
age_bins = [0, 29, 39, 49, 59, 100]\nage_labels = ['Under 30', '30-39', '40-49', '50-59', '60+']\ndf['AgeGroup'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels)\n\nage_summary = df.groupby('AgeGroup', observed=True).agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\nage_summary['churn_rate'] *= 100\nage_summary
```

```python
age_summary.plot(x='AgeGroup', y='churn_rate', kind='bar', legend=False)\nplt.title('Churn Rate by Age Group')\nplt.xlabel('Age Group')\nplt.ylabel('Churn Rate (%)')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 9. Credit Score Analysis

```python
credit_bins = [0, 499, 599, 699, 799, 900]\ncredit_labels = ['Below 500', '500-599', '600-699', '700-799', '800+']\ndf['CreditScoreGroup'] = pd.cut(\n    df['CreditScore'], bins=credit_bins, labels=credit_labels\n)\n\ncredit_summary = df.groupby('CreditScoreGroup', observed=True).agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\ncredit_summary['churn_rate'] *= 100\ncredit_summary
```

```python
credit_summary.plot(x='CreditScoreGroup', y='churn_rate', kind='bar', legend=False)\nplt.title('Churn Rate by Credit Score Group')\nplt.xlabel('Credit Score Group')\nplt.ylabel('Churn Rate (%)')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 10. Tenure Analysis

```python
tenure_summary = df.groupby('Tenure').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\ntenure_summary['churn_rate'] *= 100\ntenure_summary
```

```python
tenure_summary.plot(x='Tenure', y='churn_rate', kind='line', marker='o', legend=False)\nplt.title('Churn Rate by Tenure')\nplt.xlabel('Tenure (Years)')\nplt.ylabel('Churn Rate (%)')\nplt.grid(alpha=0.3)\nplt.tight_layout()\nplt.show()
```

## 11. Number of Products Analysis

```python
product_summary = df.groupby('NumOfProducts').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\nproduct_summary['churn_rate'] *= 100\nproduct_summary
```

```python
product_summary.plot(x='NumOfProducts', y='churn_rate', kind='bar', legend=False)\nplt.title('Churn Rate by Number of Products')\nplt.xlabel('Number of Products')\nplt.ylabel('Churn Rate (%)')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 12. Active Member Analysis

```python
active_summary = df.groupby('IsActiveMember').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\nactive_summary['MembershipStatus'] = active_summary['IsActiveMember'].map({0: 'Inactive', 1: 'Active'})\nactive_summary['churn_rate'] *= 100\nactive_summary[['MembershipStatus', 'total_customers', 'churned_customers', 'churn_rate']]
```

```python
plot_data = active_summary.set_index('MembershipStatus')['churn_rate']\nplot_data.plot(kind='bar', legend=False)\nplt.title('Churn Rate: Active vs Inactive Members')\nplt.xlabel('Membership Status')\nplt.ylabel('Churn Rate (%)')\nplt.xticks(rotation=0)\nplt.tight_layout()\nplt.show()
```

## 13. Credit Card Analysis

```python
card_summary = df.groupby('HasCrCard').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\ncard_summary['CreditCardStatus'] = card_summary['HasCrCard'].map({0: 'No Credit Card', 1: 'Has Credit Card'})\ncard_summary['churn_rate'] *= 100\ncard_summary[['CreditCardStatus', 'total_customers', 'churned_customers', 'churn_rate']]
```

## 14. Balance Analysis

```python
balance_summary = df.groupby('Exited')['Balance'].agg(['count', 'mean', 'median', 'min', 'max']).reset_index()\nbalance_summary['CustomerStatus'] = balance_summary['Exited'].map({0: 'Retained', 1: 'Churned'})\nbalance_summary[['CustomerStatus', 'count', 'mean', 'median', 'min', 'max']]
```

```python
df.boxplot(column='Balance', by='Exited')\nplt.title('Balance Distribution by Churn Status')\nplt.suptitle('')\nplt.xlabel('Exited (0 = Retained, 1 = Churned)')\nplt.ylabel('Balance')\nplt.tight_layout()\nplt.show()
```

## 15. Salary Analysis

```python
salary_summary = df.groupby('Exited')['EstimatedSalary'].agg(['count', 'mean', 'median', 'min', 'max']).reset_index()\nsalary_summary['CustomerStatus'] = salary_summary['Exited'].map({0: 'Retained', 1: 'Churned'})\nsalary_summary[['CustomerStatus', 'count', 'mean', 'median', 'min', 'max']]
```

## 16. Multi-Dimensional Analysis

```python
geo_active = df.groupby(['Geography', 'IsActiveMember']).agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\ngeo_active['MembershipStatus'] = geo_active['IsActiveMember'].map({0: 'Inactive', 1: 'Active'})\ngeo_active['churn_rate'] *= 100\ngeo_active
```

```python
age_active = df.groupby(['AgeGroup', 'IsActiveMember'], observed=True).agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\nage_active['MembershipStatus'] = age_active['IsActiveMember'].map({0: 'Inactive', 1: 'Active'})\nage_active['churn_rate'] *= 100\nage_active
```

## 17. Analytical Segment\n\nThe following is a portfolio-style segmentation rule. It is an **analytical grouping**, not evidence that these characteristics cause churn.

```python
df['ReviewSegment'] = np.where(\n    (df['IsActiveMember'] == 0) &\n    (df['Age'] >= 40) &\n    (df['NumOfProducts'] == 1),\n    'Review Segment',\n    'Other'\n)\n\nsegment_summary = df.groupby('ReviewSegment').agg(\n    total_customers=('CustomerId', 'count'),\n    churned_customers=('Exited', 'sum'),\n    churn_rate=('Exited', 'mean')\n).reset_index()\nsegment_summary['churn_rate'] *= 100\nsegment_summary
```

## 18. Correlation Analysis\n\nCorrelation can show association between numeric variables. It should not be interpreted as proof of causation.

```python
numeric_cols = [\n    'CreditScore', 'Age', 'Tenure', 'Balance',\n    'NumOfProducts', 'HasCrCard',\n    'IsActiveMember', 'EstimatedSalary', 'Exited'\n]\n\ncorrelation = df[numeric_cols].corr()\ncorrelation['Exited'].sort_values(ascending=False)
```

## 19. Export Analysis Tables\n\nThe following saves the main summary tables so they can be reused in Excel or Power BI.

```python
with pd.ExcelWriter('Bank_Churn_EDA_Summary.xlsx', engine='openpyxl') as writer:\n    geo_summary.to_excel(writer, sheet_name='Geography', index=False)\n    gender_summary.to_excel(writer, sheet_name='Gender', index=False)\n    age_summary.to_excel(writer, sheet_name='Age', index=False)\n    credit_summary.to_excel(writer, sheet_name='CreditScore', index=False)\n    tenure_summary.to_excel(writer, sheet_name='Tenure', index=False)\n    product_summary.to_excel(writer, sheet_name='Products', index=False)\n    active_summary.to_excel(writer, sheet_name='ActiveMember', index=False)\n    card_summary.to_excel(writer, sheet_name='CreditCard', index=False)\n    segment_summary.to_excel(writer, sheet_name='Segments', index=False)\n\nprint('Saved: Bank_Churn_EDA_Summary.xlsx')
```
